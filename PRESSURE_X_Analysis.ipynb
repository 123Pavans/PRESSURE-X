```python
df = ball.copy()

df['total_runs'] = df['runs'] + df['extras_2_runs']

# Wide and Overstep deliveries are not legal balls
df['legal_ball'] = ~df['extras_type'].isin(['wide', 'Overstep'])

# Sort deliveries
df = df.sort_values(
    ['match_id', 'innings', 'overid', 'over_ball_num']
).reset_index(drop=True)

# Calculate match situation BEFORE each delivery
groups = df.groupby(['match_id', 'innings'])

df['score_before'] = groups['total_runs'].cumsum() - df['total_runs']

df['wickets_before'] = (
    groups['is_wicket'].cumsum() - df['is_wicket'].astype(int)
)

df['legal_balls_before'] = (
    groups['legal_ball'].cumsum().astype(int)
    - df['legal_ball'].astype(int)
)

df['wickets_remaining'] = 10 - df['wickets_before']

print(df[[
    'match_id', 'innings', 'overid',
    'score_before', 'wickets_before',
    'legal_balls_before', 'total_runs'
]].head(15))
```

```python
first_innings = (
    df[df['innings'] == 1]
    .groupby('match_id')['total_runs']
    .sum()
    .add(1)
)

df['target'] = df['match_id'].map(first_innings)

# Use revised target where available
revised = matches.set_index('match_id')['revised_target']

df['revised_target'] = df['match_id'].map(revised)

df['target'] = df['revised_target'].fillna(df['target'])

# Standard T20 assumption: 20 overs
df['balls_limit'] = 120

# Account for revised innings duration
for col in ['first_inn_revised_overs', 'second_inn_revised_overs']:
    pass

df.loc[df['innings'] == 2, 'balls_remaining'] = (
    df.loc[df['innings'] == 2, 'balls_limit']
    - df.loc[df['innings'] == 2, 'legal_balls_before']
)

df['runs_required'] = df['target'] - df['score_before']

df['overs_remaining'] = df['balls_remaining'] / 6

df['required_run_rate'] = np.where(
    (df['overs_remaining'] > 0) & (df['innings'] == 2),
    df['runs_required'] / df['overs_remaining'],
    np.nan
)
```

```python
print("1. Delivery data:")
print(df.shape)

print("\n2. Missing values:")
print(df[['total_runs', 'score_before',
          'wickets_before', 'legal_balls_before',
          'target', 'balls_remaining',
          'required_run_rate']].isnull().sum())

print("\n3. Second innings sample:")
display(
    df[df['innings'] == 2][[
        'match_id', 'overid', 'score_before',
        'wickets_before', 'target',
        'runs_required', 'balls_remaining',
        'required_run_rate'
    ]].sample(10, random_state=42)
)

print("\n4. Negative runs required:")
print((df['runs_required'] < 0).sum())

print("\n5. Negative balls remaining:")
print((df['balls_remaining'] < 0).sum())
```

```python
# Get innings limits from match metadata
df['first_overs'] = df['match_id'].map(
    matches.set_index('match_id')['first_inn_revised_overs']
)

df['second_overs'] = df['match_id'].map(
    matches.set_index('match_id')['second_inn_revised_overs']
)

# Default to 20 overs where no revised limit exists
df['first_overs'] = df['first_overs'].fillna(20)
df['second_overs'] = df['second_overs'].fillna(20)

df['innings_limit'] = np.where(
    df['innings'] == 1,
    df['first_overs'] * 6,
    df['second_overs'] * 6
)

df['balls_remaining'] = (
    df['innings_limit'] - df['legal_balls_before']
).clip(lower=0)

# Recalculate runs required
df['runs_required'] = df['target'] - df['score_before']

df['required_run_rate'] = np.where(
    (df['innings'] == 2) &
    (df['balls_remaining'] > 0) &
    (df['runs_required'] > 0),
    df['runs_required'] * 6 / df['balls_remaining'],
    np.nan
)
```

```python
df['active_chase'] = (
    (df['innings'] == 2) &
    (df['runs_required'] > 0) &
    (df['balls_remaining'] > 0) &
    (df['wickets_remaining'] > 0)
)

print("Total deliveries:", len(df))

print("Active chase deliveries:",
      df['active_chase'].sum())

print("Excluded deliveries:",
      (~df['active_chase']).sum())
```

```python
# Create revised-overs and active-chase features if not already created
df['first_overs'] = df['match_id'].map(
    matches.set_index('match_id')['first_inn_revised_overs']
).fillna(20)

df['second_overs'] = df['match_id'].map(
    matches.set_index('match_id')['second_inn_revised_overs']
).fillna(20)

df['innings_limit'] = np.where(
    df['innings'] == 1,
    df['first_overs'] * 6,
    df['second_overs'] * 6
)

df['balls_remaining'] = (
    df['innings_limit'] - df['legal_balls_before']
).clip(lower=0)

df['runs_required'] = df['target'] - df['score_before']

df['required_run_rate'] = np.where(
    (df['innings'] == 2) &
    (df['balls_remaining'] > 0) &
    (df['runs_required'] > 0),
    df['runs_required'] * 6 / df['balls_remaining'],
    np.nan
)

df['active_chase'] = (
    (df['innings'] == 2) &
    (df['runs_required'] > 0) &
    (df['balls_remaining'] > 0) &
    (df['wickets_remaining'] > 0)
)

# Validation
print("Total deliveries:", len(df))
print("Active chase deliveries:", df['active_chase'].sum())
print("Unique chase matches:", df.loc[df['active_chase'], 'match_id'].nunique())
print("\nInnings distribution:")
print(df['innings'].value_counts().sort_index())
print("\nMissing target values:", df['target'].isna().sum())
print("Missing required run rate in active chases:",
      df.loc[df['active_chase'], 'required_run_rate'].isna().sum())
```

```python
chase = df[df['active_chase']].copy()

# Normalize pressure factors between 0 and 1
chase['rrr_pressure'] = (
    chase['required_run_rate'] / 15
).clip(0, 1)

chase['wicket_pressure'] = (
    10 - chase['wickets_remaining']
) / 10

chase['time_pressure'] = (
    1 - chase['balls_remaining'] /
    chase['innings_limit']
).clip(0, 1)

# Dynamic Pressure Index
chase['DPI'] = 100 * (
    0.40 * chase['rrr_pressure'] +
    0.35 * chase['wicket_pressure'] +
    0.25 * chase['time_pressure']
)

# Pressure categories
chase['pressure_level'] = pd.cut(
    chase['DPI'],
    bins=[-1, 35, 60, 100],
    labels=['Low', 'Medium', 'High']
)

print(chase['pressure_level'].value_counts())
print("\nDPI Summary:")
print(chase['DPI'].describe())
```

```python
high_pressure = chase[
    chase['pressure_level'] == 'High'
].copy()

print("High-pressure deliveries:", len(high_pressure))
print("Matches:", high_pressure['match_id'].nunique())

print("\nTop batters by high-pressure balls faced:")
print(
    high_pressure.groupby('striker_name')
    .agg(
        balls_faced=('striker_id', 'count'),
        runs=('runs', 'sum')
    )
    .sort_values('balls_faced', ascending=False)
    .head(10)
)
```

```python
batter_pressure = high_pressure.groupby(
    ['striker_id', 'striker_name']
).agg(
    balls_faced=('striker_id', 'count'),
    runs_scored=('runs', 'sum'),
    dismissals=('is_wicket', 'sum'),
    fours=('is_four', 'sum'),
    sixes=('is_six', 'sum')
).reset_index()

batter_pressure['strike_rate'] = (
    batter_pressure['runs_scored'] /
    batter_pressure['balls_faced'] * 100
)

batter_pressure['boundary_runs'] = (
    batter_pressure['fours'] * 4 +
    batter_pressure['sixes'] * 6
)

batter_pressure = batter_pressure.sort_values(
    'strike_rate', ascending=False
)

print(batter_pressure.head(20).round(2))
```

```python
eligible_batters = batter_pressure[
    batter_pressure['balls_faced'] >= 30
].copy()

eligible_batters = eligible_batters.sort_values(
    'strike_rate', ascending=False
)

print("Eligible batters:", len(eligible_batters))
print(
    eligible_batters[
        ['striker_name', 'balls_faced', 'runs_scored',
         'dismissals', 'strike_rate']
    ].head(20).round(2)
)
```

```python
normal_batting = df[df['innings'].isin([1, 2])].groupby(
    ['striker_id', 'striker_name']
).agg(
    total_balls=('striker_id', 'count'),
    total_runs=('runs', 'sum'),
    total_dismissals=('is_wicket', 'sum')
).reset_index()

normal_batting['baseline_sr'] = (
    normal_batting['total_runs'] /
    normal_batting['total_balls'] * 100
)

batter_comparison = eligible_batters.merge(
    normal_batting[
        ['striker_id', 'baseline_sr', 'total_balls']
    ],
    on='striker_id',
    how='left'
)

batter_comparison['sr_difference'] = (
    batter_comparison['strike_rate'] -
    batter_comparison['baseline_sr']
)

print(
    batter_comparison[
        ['striker_name', 'balls_faced', 'strike_rate',
         'baseline_sr', 'sr_difference']
    ].sort_values('sr_difference', ascending=False)
    .head(20).round(2)
)
```

```python
# Exclude wides when counting balls faced
high_pressure['batter_ball'] = (
    high_pressure['extras_type'] != 'wide'
)

baseline_deliveries = chase[
    chase['pressure_level'].isin(['Low', 'Medium'])
].copy()

baseline_deliveries['batter_ball'] = (
    baseline_deliveries['extras_type'] != 'wide'
)

# Pressure performance
pressure_stats = high_pressure.groupby(
    ['striker_id', 'striker_name']
).agg(
    pressure_balls=('batter_ball', 'sum'),
    pressure_runs=('runs', 'sum'),
    pressure_dismissals=('is_wicket', 'sum')
).reset_index()

pressure_stats['pressure_sr'] = (
    pressure_stats['pressure_runs'] /
    pressure_stats['pressure_balls'] * 100
)

# Non-high-pressure baseline
baseline_stats = baseline_deliveries.groupby(
    'striker_id'
).agg(
    baseline_balls=('batter_ball', 'sum'),
    baseline_runs=('runs', 'sum')
).reset_index()

baseline_stats['baseline_sr'] = (
    baseline_stats['baseline_runs'] /
    baseline_stats['baseline_balls'] * 100
)

# Compare
batter_comparison = pressure_stats.merge(
    baseline_stats,
    on='striker_id',
    how='inner'
)

batter_comparison = batter_comparison[
    (batter_comparison['pressure_balls'] >= 30) &
    (batter_comparison['baseline_balls'] >= 30)
].copy()

batter_comparison['sr_difference'] = (
    batter_comparison['pressure_sr'] -
    batter_comparison['baseline_sr']
)

print("Eligible batters:", len(batter_comparison))

print(
    batter_comparison.sort_values(
        'sr_difference', ascending=False
    )[
        ['striker_name', 'pressure_balls', 'pressure_sr',
         'baseline_sr', 'sr_difference']
    ].head(20).round(2)
)
```

```python
batter_comparison['reliability'] = (
    batter_comparison['pressure_balls'] / 100
).clip(upper=1)

batter_comparison['adjusted_improvement'] = (
    batter_comparison['sr_difference'] *
    batter_comparison['reliability']
)

top_batters = batter_comparison.sort_values(
    'adjusted_improvement', ascending=False
)

print(
    top_batters[
        ['striker_name', 'pressure_balls', 'pressure_sr',
         'baseline_sr', 'sr_difference',
         'reliability', 'adjusted_improvement']
    ].head(20).round(2).to_string(index=False)
)
```

```python
top_batters = top_batters.merge(
    baseline_stats[['striker_id', 'baseline_balls']],
    on='striker_id',
    how='left',
    suffixes=('', '_check')
)

print(
    top_batters[
        ['striker_name', 'pressure_balls',
         'baseline_balls', 'pressure_sr',
         'baseline_sr', 'adjusted_improvement']
    ].head(20).round(2).to_string(index=False)
)
```

```python
# Require sufficient baseline data
reliable_batters = batter_comparison[
    (batter_comparison['pressure_balls'] >= 30) &
    (batter_comparison['baseline_balls'] >= 100)
].copy()

# Pressure reliability
reliable_batters['pressure_reliability'] = (
    reliable_batters['pressure_balls'] / 100
).clip(upper=1)

# Baseline reliability
reliable_batters['baseline_reliability'] = (
    reliable_batters['baseline_balls'] / 200
).clip(upper=1)

# Combined reliability
reliable_batters['combined_reliability'] = np.sqrt(
    reliable_batters['pressure_reliability'] *
    reliable_batters['baseline_reliability']
)

# Final adjusted score
reliable_batters['final_score'] = (
    reliable_batters['sr_difference'] *
    reliable_batters['combined_reliability']
)

top_batters = reliable_batters.sort_values(
    'final_score', ascending=False
)

print("Reliable eligible batters:", len(top_batters))

print(
    top_batters[
        ['striker_name', 'pressure_balls', 'baseline_balls',
         'pressure_sr', 'baseline_sr',
         'sr_difference', 'combined_reliability',
         'final_score']
    ].head(20).round(2).to_string(index=False)
)
```

```python
top_batters['match_reliability'] = (
    top_batters['pressure_matches'] / 15
).clip(upper=1)

top_batters['final_batting_score'] = (
    top_batters['final_score'] *
    (0.7 + 0.3 * top_batters['match_reliability'])
)

final_batter_ranking = top_batters.sort_values(
    'final_batting_score', ascending=False
)

print(
    final_batter_ranking[
        ['striker_name', 'pressure_balls',
         'pressure_matches', 'final_score',
         'final_batting_score']
    ].head(15).round(2).to_string(index=False)
)
```

```python
# Separate high-pressure and baseline deliveries
high_bowling = high_pressure.copy()

baseline_bowling = chase[
    chase['pressure_level'].isin(['Low', 'Medium'])
].copy()

# Exclude byes and leg-byes from bowler runs conceded
def bowler_runs(row):
    extras = str(row['extras_type']).lower()
    if 'bye' in extras or 'penalty' in extras:
        return row['runs']
    return row['runs'] + row['extras_2_runs']

high_bowling['runs_conceded'] = high_bowling.apply(bowler_runs, axis=1)
baseline_bowling['runs_conceded'] = baseline_bowling.apply(bowler_runs, axis=1)

# Wickets credited to the bowler
bowler_dismissals = r'bowled|caught|lbw|stumped|hit.?wicket'

for data in [high_bowling, baseline_bowling]:
    data['bowler_wicket'] = (
        (data['is_wicket'] == 1) &
        data['dismissal_type'].str.contains(
            bowler_dismissals, case=False, na=False
        )
    ).astype(int)

print("High-pressure bowling deliveries:", len(high_bowling))
print("Baseline bowling deliveries:", len(baseline_bowling))
```

```python
def bowling_stats(data, prefix):
    return data.groupby(['bowler_id', 'bowler_name']).agg(
        **{
            f'{prefix}_balls': ('legal_ball', 'sum'),
            f'{prefix}_runs': ('runs_conceded', 'sum'),
            f'{prefix}_wickets': ('bowler_wicket', 'sum'),
            f'{prefix}_matches': ('match_id', 'nunique')
        }
    ).reset_index()

pressure_bowl_stats = bowling_stats(high_bowling, 'pressure')
baseline_bowl_stats = bowling_stats(baseline_bowling, 'baseline')

bowler_comparison = pressure_bowl_stats.merge(
    baseline_bowl_stats,
    on='bowler_id',
    how='inner',
    suffixes=('', '_baseline')
)

bowler_comparison['pressure_economy'] = (
    bowler_comparison['pressure_runs'] /
    bowler_comparison['pressure_balls'] * 6
)

bowler_comparison['baseline_economy'] = (
    bowler_comparison['baseline_runs'] /
    bowler_comparison['baseline_balls'] * 6
)

bowler_comparison['pressure_sr'] = np.where(
    bowler_comparison['pressure_wickets'] > 0,
    bowler_comparison['pressure_balls'] /
    bowler_comparison['pressure_wickets'],
    np.nan
)

bowler_comparison['baseline_sr'] = np.where(
    bowler_comparison['baseline_wickets'] > 0,
    bowler_comparison['baseline_balls'] /
    bowler_comparison['baseline_wickets'],
    np.nan
)

display(bowler_comparison.head())
```

```python
# Minimum sample requirements
eligible_bowlers = bowler_comparison[
    (bowler_comparison['pressure_balls'] >= 60) &
    (bowler_comparison['baseline_balls'] >= 120) &
    (bowler_comparison['pressure_matches'] >= 5) &
    (bowler_comparison['baseline_matches'] >= 5) &
    (bowler_comparison['pressure_wickets'] >= 5) &
    (bowler_comparison['baseline_wickets'] >= 5)
].copy()

# Positive values indicate improvement under pressure
eligible_bowlers['economy_improvement'] = (
    (eligible_bowlers['baseline_economy'] -
     eligible_bowlers['pressure_economy']) /
    eligible_bowlers['baseline_economy'] * 100
)

eligible_bowlers['wicket_improvement'] = (
    (eligible_bowlers['baseline_sr'] -
     eligible_bowlers['pressure_sr']) /
    eligible_bowlers['baseline_sr'] * 100
)

# Combined performance score
eligible_bowlers['performance_score'] = (
    0.5 * eligible_bowlers['economy_improvement'] +
    0.5 * eligible_bowlers['wicket_improvement']
)

# Sample reliability adjustment
eligible_bowlers['reliability'] = np.sqrt(
    (eligible_bowlers['pressure_balls'] / 120).clip(upper=1) *
    (eligible_bowlers['baseline_balls'] / 300).clip(upper=1)
)

eligible_bowlers['final_bowling_score'] = (
    eligible_bowlers['performance_score'] *
    eligible_bowlers['reliability']
)

print("Eligible bowlers:", len(eligible_bowlers))

display(
    eligible_bowlers.sort_values(
        'final_bowling_score', ascending=False
    )[[
        'bowler_name',
        'pressure_balls',
        'pressure_matches',
        'pressure_economy',
        'baseline_economy',
        'economy_improvement',
        'pressure_sr',
        'baseline_sr',
        'wicket_improvement',
        'reliability',
        'final_bowling_score'
    ]].head(20)
)
```

```python
df = df.sort_values(
    ['match_id', 'innings', 'overid', 'over_ball_num']
).reset_index(drop=True)

groups = df.groupby(['match_id', 'innings'])

df['score_before'] = (
    groups['total_runs'].cumsum() - df['total_runs']
)

df['wickets_before'] = (
    groups['is_wicket'].cumsum() - df['is_wicket'].astype(int)
)

df['legal_ball'] = ~df['extras_type'].isin(['wide', 'Overstep'])

df['legal_balls_before'] = (
    groups['legal_ball'].cumsum().astype(int)
    - df['legal_ball'].astype(int)
)

df['wickets_remaining'] = 10 - df['wickets_before']

first_innings = (
    df[df['innings'] == 1]
    .groupby('match_id')['total_runs']
    .sum() + 1
)

df['target'] = df['match_id'].map(first_innings)

revised = matches.set_index('match_id')['revised_target']
df['target'] = df['match_id'].map(revised).fillna(df['target'])

df['first_overs'] = df['match_id'].map(
    matches.set_index('match_id')['first_inn_revised_overs']
).fillna(20)

df['second_overs'] = df['match_id'].map(
    matches.set_index('match_id')['second_inn_revised_overs']
).fillna(20)

df['innings_limit'] = np.where(
    df['innings'] == 1,
    df['first_overs'] * 6,
    df['second_overs'] * 6
)

df['balls_remaining'] = (
    df['innings_limit'] - df['legal_balls_before']
).clip(lower=0)

df['runs_required'] = df['target'] - df['score_before']

df['required_run_rate'] = np.where(
    (df['innings'] == 2) &
    (df['balls_remaining'] > 0) &
    (df['runs_required'] > 0),
    df['runs_required'] * 6 / df['balls_remaining'],
    np.nan
)

df['active_chase'] = (
    (df['innings'] == 2) &
    (df['runs_required'] > 0) &
    (df['balls_remaining'] > 0) &
    (df['wickets_remaining'] > 0)
)
```

```python
print("Active chase deliveries:", df['active_chase'].sum())
print("Missing targets:", df['target'].isna().sum())
print("Missing RRR:", df.loc[df['active_chase'], 'required_run_rate'].isna().sum())
```

```python
chase = df[df['active_chase']].copy()

chase['rrr_pressure'] = (
    chase['required_run_rate'] / 15
).clip(0, 1)

chase['wicket_pressure'] = (
    10 - chase['wickets_remaining']
) / 10

chase['time_pressure'] = (
    1 - chase['balls_remaining'] / chase['innings_limit']
).clip(0, 1)

chase['DPI'] = 100 * (
    0.40 * chase['rrr_pressure'] +
    0.35 * chase['wicket_pressure'] +
    0.25 * chase['time_pressure']
)

chase['pressure_level'] = pd.cut(
    chase['DPI'],
    bins=[-1, 35, 60, 100],
    labels=['Low', 'Medium', 'High']
)

high_pressure = chase[
    chase['pressure_level'] == 'High'
].copy()

print(chase['pressure_level'].value_counts())
print("High-pressure deliveries:", len(high_pressure))
print("High-pressure matches:", high_pressure['match_id'].nunique())
print("Mean DPI:", chase['DPI'].mean())
```

```python
high_pressure['batter_ball'] = (
    high_pressure['extras_type'] != 'wide'
)

baseline_deliveries = chase[
    chase['pressure_level'].isin(['Low', 'Medium'])
].copy()

baseline_deliveries['batter_ball'] = (
    baseline_deliveries['extras_type'] != 'wide'
)

pressure_stats = high_pressure.groupby(
    ['striker_id', 'striker_name']
).agg(
    pressure_balls=('batter_ball', 'sum'),
    pressure_runs=('runs', 'sum'),
    pressure_dismissals=('is_wicket', 'sum')
).reset_index()

pressure_stats['pressure_sr'] = (
    pressure_stats['pressure_runs'] /
    pressure_stats['pressure_balls'] * 100
)

baseline_stats = baseline_deliveries.groupby(
    'striker_id'
).agg(
    baseline_balls=('batter_ball', 'sum'),
    baseline_runs=('runs', 'sum')
).reset_index()

baseline_stats['baseline_sr'] = (
    baseline_stats['baseline_runs'] /
    baseline_stats['baseline_balls'] * 100
)

batter_comparison = pressure_stats.merge(
    baseline_stats,
    on='striker_id',
    how='inner'
)

batter_comparison = batter_comparison[
    (batter_comparison['pressure_balls'] >= 30) &
    (batter_comparison['baseline_balls'] >= 100)
].copy()

batter_comparison['sr_difference'] = (
    batter_comparison['pressure_sr'] -
    batter_comparison['baseline_sr']
)

print("Eligible batters:", len(batter_comparison))
```

```python
reliable_batters = batter_comparison.copy()

reliable_batters['pressure_reliability'] = (
    reliable_batters['pressure_balls'] / 100
).clip(upper=1)

reliable_batters['baseline_reliability'] = (
    reliable_batters['baseline_balls'] / 200
).clip(upper=1)

reliable_batters['combined_reliability'] = np.sqrt(
    reliable_batters['pressure_reliability'] *
    reliable_batters['baseline_reliability']
)

reliable_batters['final_score'] = (
    reliable_batters['sr_difference'] *
    reliable_batters['combined_reliability']
)

final_batter_ranking = reliable_batters.sort_values(
    'final_score', ascending=False
)

display(
    final_batter_ranking[
        [
            'striker_name',
            'pressure_balls',
            'pressure_sr',
            'baseline_sr',
            'sr_difference',
            'final_score'
        ]
    ].head(15)
)
```

```python
high_bowling = high_pressure.copy()

baseline_bowling = chase[
    chase['pressure_level'].isin(['Low', 'Medium'])
].copy()

for data in [high_bowling, baseline_bowling]:
    data['bowler_wicket'] = (
        (data['is_wicket'] == 1) &
        data['dismissal_type'].str.contains(
            r'bowled|caught|lbw|stump|hit.?wicket',
            case=False,
            na=False
        )
    ).astype(int)

def bowling_stats(data, prefix):
    return data.groupby(
        ['bowler_id', 'bowler_name']
    ).agg(
        **{
            f'{prefix}_balls': ('legal_ball', 'sum'),
            f'{prefix}_runs': ('bowler_runs', 'sum'),
            f'{prefix}_wickets': ('bowler_wicket', 'sum'),
            f'{prefix}_matches': ('match_id', 'nunique')
        }
    ).reset_index()

pressure_bowl_stats = bowling_stats(
    high_bowling, 'pressure'
)

baseline_bowl_stats = bowling_stats(
    baseline_bowling, 'baseline'
)

bowler_comparison = pressure_bowl_stats.merge(
    baseline_bowl_stats,
    on='bowler_id',
    suffixes=('', '_baseline')
)

print("Eligible bowling comparisons:", len(bowler_comparison))
display(bowler_comparison.head())
```

```python
bowler_comparison['pressure_economy'] = (
    bowler_comparison['pressure_runs'] /
    bowler_comparison['pressure_balls'] * 6
)

bowler_comparison['baseline_economy'] = (
    bowler_comparison['baseline_runs'] /
    bowler_comparison['baseline_balls'] * 6
)

bowler_comparison['pressure_sr'] = np.where(
    bowler_comparison['pressure_wickets'] > 0,
    bowler_comparison['pressure_balls'] /
    bowler_comparison['pressure_wickets'],
    np.nan
)

bowler_comparison['baseline_sr'] = np.where(
    bowler_comparison['baseline_wickets'] > 0,
    bowler_comparison['baseline_balls'] /
    bowler_comparison['baseline_wickets'],
    np.nan
)
```

```python
eligible_bowlers = bowler_comparison[
    (bowler_comparison['pressure_balls'] >= 60) &
    (bowler_comparison['baseline_balls'] >= 120) &
    (bowler_comparison['pressure_matches'] >= 5) &
    (bowler_comparison['baseline_matches'] >= 5) &
    (bowler_comparison['pressure_wickets'] >= 5) &
    (bowler_comparison['baseline_wickets'] >= 5)
].copy()

print("Eligible bowlers:", len(eligible_bowlers))
```

```python
eligible_bowlers['economy_improvement'] = (
    (eligible_bowlers['baseline_economy'] -
     eligible_bowlers['pressure_economy']) /
    eligible_bowlers['baseline_economy'] * 100
)

eligible_bowlers['wicket_improvement'] = (
    (eligible_bowlers['baseline_sr'] -
     eligible_bowlers['pressure_sr']) /
    eligible_bowlers['baseline_sr'] * 100
)

eligible_bowlers['performance_score'] = (
    0.5 * eligible_bowlers['economy_improvement'] +
    0.5 * eligible_bowlers['wicket_improvement']
)

eligible_bowlers['reliability'] = np.sqrt(
    (eligible_bowlers['pressure_balls'] / 120).clip(upper=1) *
    (eligible_bowlers['baseline_balls'] / 300).clip(upper=1)
)

eligible_bowlers['final_bowling_score'] = (
    eligible_bowlers['performance_score'] *
    eligible_bowlers['reliability']
)

final_bowler_ranking = eligible_bowlers.sort_values(
    'final_bowling_score',
    ascending=False
)

display(final_bowler_ranking.head(15))
```
